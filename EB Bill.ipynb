```python
Consumer_Name = input("Enter Consumer Name: ")
Tariff = input("Enter Tarrif Type (Domestic/Commercial): ")
Consumed_Units = int(input("Enter Consumed Units: "))
Subsidy = 0
New_Subsidy = 250

if Tariff == "Domestic":
  if Consumed_Units <= 100:
    Total_Current_Charges = Consumed_Units * 3
    Subsidy = 25
  elif Consumed_Units <= 200:
    Total_Current_Charges = Consumed_Units * 5
    Subsidy = 50
  else:
    Total_Current_Charges = Consumed_Units * 7
    Subsidy = 75
elif Tariff == "Commercial":
  if Consumed_Units <= 100:
    Total_Current_Charges = Consumed_Units * 5
    E_Tax = 100
  elif Consumed_Units <= 200:
    Total_Current_Charges = Consumed_Units * 8
    E_Tax = 200
  else:
    Total_Current_Charges = Consumed_Units * 10
    E_Tax = 300

if Tariff == "Domestic":
  Net_Charges = Total_Current_Charges - Subsidy - New_Subsidy
else:
  Net_Charges = Total_Current_Charges + E_Tax

print("\n        EB BILL        ")
print("Name of the Consumer: ", Consumer_Name)
print("Tariff Type: ", Tariff)
print("Consumed Units: ", Consumed_Units)
print("Total Current Charges: ", Total_Current_Charges)
if Tariff == "Domestic":
  print("Subsidy: ", Subsidy)
  print("New Subsidy: ", New_Subsidy)
else:
  print("Extra Tax: ", E_Tax)
print("Net Charges: ", Net_Charges)

```
